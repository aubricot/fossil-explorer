## 🤖 Mission 2: Ask the AI
The AI turns every 3D shape into a list of numbers called a **shape code**. Bones with similar shapes get similar shape codes.

To identify a fossil, the AI compares the fossil's shape code with the codes of every modern bone it knows, and reports the **5 most similar**.

```python
matches = find_matches("mystery_fossil")
```

Now put the fossil **side by side** with the AI's #1 match.

## 🔨 Mission 5: Break a bone and let the AI fix it
Let's test the AI on a bone where **we know exactly what the answer should be**. We'll take a modern bone, break off a piece, and see whether the AI can rebuild it *and still recognize it*.

```python
bone = pick_modern_bone()
```

### Step B: Break it! 🔨
* `how_much` is how much to chop off: `0.1` is a small chip, `0.5` is half the bone.

```python
broken = break_bone(bone, how_much=0.3, side="top")
```
